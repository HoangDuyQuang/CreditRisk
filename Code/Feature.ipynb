```python
import pandas as pd
import numpy as np 
```

```python
df = pd.read_csv('Data.csv')
```

```python
df.info()
```

```python
# Free cash flow after debt repayment 
df['free_cash_flow'] = df['person_income'] * (1 - df['loan_percent_income'])

# Cash flow calculated based on total debt.
df['net_income_after_total_debt'] = df['person_income'] * (1 - df['debt_to_income_ratio'])

# Punish those with high debt but zero years of service
df['risk_emp_debt'] = df['loan_percent_income'] / (df['person_emp_length'] + 1)

# Loan-to-income ratio adjusted for income
df['adjusted_loan_to_income'] = df['loan_to_income_ratio'] / np.log1p(df['person_income'])
```

```python
df.to_csv('Data.csv',index=False, encoding="utf-8-sig")
```
